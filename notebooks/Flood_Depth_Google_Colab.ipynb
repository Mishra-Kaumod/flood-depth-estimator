# Flood Depth Estimator — Balanced Training + HPO (Colab)

**Pipeline:**
1. Load labels.csv + images, audit alignment
2. Visualise depth distribution and diagnose imbalance
3. Balance the training set (augmentation for rare bins, cap for dominant bins)
4. Extract EfficientNet-B0 features
5. Hyperparameter search across Ridge / Lasso / ElasticNet / SVR with cross-validation
6. Fine-tune EfficientNet head with early stopping + cosine LR
7. Evaluate on held-out test set, plot residuals, save model


```python
# ── Colab: uncomment if packages are missing ──────────────────────────────────
# %pip install -q efficientnet_pytorch scikit-learn pandas matplotlib tqdm Pillow torchvision
```

```python
import os, warnings, random
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance, ImageOps
from tqdm.auto import tqdm

import torch
from torchvision import transforms, models
import torch.nn as nn

from sklearn.model_selection import train_test_split, RandomizedSearchCV, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from scipy.stats import loguniform, uniform

warnings.filterwarnings('ignore')
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)

# ── Paths ──────────────────────────────────────────────────────────────────────
DATA_DIR   = TRAINING_DATA_DIR if 'TRAINING_DATA_DIR' in globals() else Path('training_data')
IMAGES_DIR = DATA_DIR / 'images'
LABELS_CSV = DATA_DIR / 'labels.csv'

assert LABELS_CSV.exists(), f'labels.csv not found at {LABELS_CSV}'
assert IMAGES_DIR.exists(), f'images dir not found at {IMAGES_DIR}'

df = pd.read_csv(LABELS_CSV)
df['depth_cm'] = df['depth_cm'].astype(float)
print(f'Loaded {len(df)} rows from labels.csv')
df.head()
```

```python
# ══ 1. Audit: images ↔ labels alignment ══════════════════════════════════════
missing = [fn for fn in df['filename'] if not (IMAGES_DIR / fn).exists()]
known   = set(df['filename'])
extra   = [p.name for p in IMAGES_DIR.iterdir() if p.is_file() and p.name not in known]

print(f'Missing files (label has no image): {len(missing)}')
if missing: print(' ', missing[:10])
print(f'Extra  files  (image has no label):  {len(extra)}')
if extra:   print(' ', extra[:10])

# Drop rows whose image file is absent
df = df[df['filename'].apply(lambda fn: (IMAGES_DIR / fn).exists())].reset_index(drop=True)
print(f'After audit: {len(df)} rows remain')
```

```python
# ══ 2. Distribution analysis ══════════════════════════════════════════════════
BINS   = [-1, 5, 20, 50, 80, 1e9]
BLABELS= ['0-5 cm','5-20 cm','20-50 cm','50-80 cm','80+ cm']

df['bin'] = pd.cut(df['depth_cm'], bins=BINS, labels=BLABELS)
counts = df['bin'].value_counts().reindex(BLABELS)

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
counts.plot(kind='bar', ax=axes[0], color='steelblue', title='Samples per depth bin')
axes[0].set_xlabel('Depth bin'); axes[0].set_ylabel('Count')
axes[1].hist(df['depth_cm'], bins=40, color='teal', edgecolor='white')
axes[1].set_title('Depth histogram (raw)'); axes[1].set_xlabel('depth_cm')
plt.tight_layout(); plt.show()

print('Counts per bin:')
print(counts.to_string())

# Imbalance ratio: max/min count
ratio = counts.max() / (counts.min() + 1e-9)
print(f'\nImbalance ratio (max/min): {ratio:.1f}')
if ratio > 3:
    print('⚠  Dataset is imbalanced — balancing will be applied.')
else:
    print('✓  Dataset looks reasonably balanced.')
```

```python
# ══ 3a. Stratified train / val / test split ═══════════════════════════════════
n = len(df)
if n >= 700:
    train_raw, temp = train_test_split(df, train_size=600, stratify=df['bin'], random_state=SEED)
    remainder = n - 600
    val_size  = min(50, remainder // 2)
    test_size = min(50, remainder - val_size)
    val,  test = train_test_split(temp, train_size=val_size, test_size=test_size,
                                  stratify=temp['bin'], random_state=SEED)
else:
    train_raw, temp = train_test_split(df, test_size=0.20, stratify=df['bin'], random_state=SEED)
    val, test = train_test_split(temp, test_size=0.50, stratify=temp['bin'], random_state=SEED)

print(f'Split  → train {len(train_raw)} | val {len(val)} | test {len(test)}')
print('Train bin distribution (before balancing):')
print(train_raw['bin'].value_counts().reindex(BLABELS))
```

```python
# ══ 3b. Balance training set ══════════════════════════════════════════════════
#
# Strategy:
#   - Compute target_per_bin = median count (robust to outlier-heavy bins)
#   - Under-represented bins → oversample with augmentation flags
#   - Over-represented bins  → downsample (random under-sampling)
#
bin_counts  = train_raw['bin'].value_counts().reindex(BLABELS).fillna(0).astype(int)
target_per_bin = int(np.median(bin_counts[bin_counts > 0]))
target_per_bin = max(target_per_bin, 30)   # floor so tiny bins grow a bit
print(f'Target per bin: {target_per_bin}')

balanced_parts = []
for lbl in BLABELS:
    grp = train_raw[train_raw['bin'] == lbl]
    have = len(grp)
    if have == 0:
        continue
    if have < target_per_bin:
        # oversample + mark augment=True for minority rows
        n_extra = target_per_bin - have
        extra   = grp.sample(n_extra, replace=True, random_state=SEED).copy()
        extra['augment'] = True
        grp = grp.copy(); grp['augment'] = False
        balanced_parts.append(pd.concat([grp, extra]))
    else:
        # downsample
        grp = grp.sample(target_per_bin, replace=False, random_state=SEED).copy()
        grp['augment'] = False
        balanced_parts.append(grp)

train_bal = pd.concat(balanced_parts).sample(frac=1, random_state=SEED).reset_index(drop=True)

print('Balanced train distribution:')
print(train_bal['bin'].value_counts().reindex(BLABELS))
print(f'Total balanced train size: {len(train_bal)}')

# quick bar chart
train_bal['bin'].value_counts().reindex(BLABELS).plot(kind='bar', color='orange', title='Balanced train distribution')
plt.tight_layout(); plt.show()
```

```python
# ══ 4. Dataset + augmentation ══════════════════════════════════════════════════
#
# Augment flag = random horizontal flip + colour jitter + small rotation
#
base_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])
aug_tf = transforms.Compose([
    transforms.Resize((240, 240)),
    transforms.RandomCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.05),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])

class FloodDataset(torch.utils.data.Dataset):
    def __init__(self, df, img_dir, train=False):
        self.df      = df.reset_index(drop=True)
        self.img_dir = Path(img_dir)
        self.train   = train
        self.aug_set = set(df.index[df.get('augment', False) == True].tolist()) if train else set()
    def __len__(self): return len(self.df)
    def __getitem__(self, idx):
        row  = self.df.iloc[idx]
        img  = Image.open(self.img_dir / row['filename']).convert('RGB')
        tf   = aug_tf if (self.train and row.get('augment', False)) else base_tf
        return tf(img), np.float32(row['depth_cm'])

train_ds = FloodDataset(train_bal, IMAGES_DIR, train=True)
val_ds   = FloodDataset(val,       IMAGES_DIR)
test_ds  = FloodDataset(test,      IMAGES_DIR)

NWORKERS = 2
train_loader = torch.utils.data.DataLoader(train_ds, batch_size=32, shuffle=True,  num_workers=NWORKERS, pin_memory=True)
val_loader   = torch.utils.data.DataLoader(val_ds,   batch_size=64, shuffle=False, num_workers=NWORKERS, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_ds,  batch_size=64, shuffle=False, num_workers=NWORKERS, pin_memory=True)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
print('Train / Val / Test dataset sizes:', len(train_ds), len(val_ds), len(test_ds))
```

```python
# ══ 5a. Feature extraction (EfficientNet-B0 pretrained, frozen) ════════════════
backbone        = models.efficientnet_b0(pretrained=True)
feat_extractor  = nn.Sequential(backbone.features, nn.AdaptiveAvgPool2d(1), nn.Flatten(1)).to(device)
feat_extractor.eval()

def extract_feats(loader, desc=''):
    Xs, ys = [], []
    with torch.no_grad():
        for imgs, labels in tqdm(loader, desc=desc):
            Xs.append(feat_extractor(imgs.to(device)).cpu().numpy())
            ys.append(labels.numpy())
    return np.vstack(Xs), np.concatenate(ys)

# For feature extraction use base transform (no augment noise)
fe_train_ds = FloodDataset(train_bal, IMAGES_DIR, train=False)   # no aug for feature pass
fe_loader   = torch.utils.data.DataLoader(fe_train_ds, batch_size=64, shuffle=False, num_workers=NWORKERS)

X_train, y_train = extract_feats(fe_loader,   'Train feats')
X_val,   y_val   = extract_feats(val_loader,  'Val feats')
X_test,  y_test  = extract_feats(test_loader, 'Test feats')
print('Feature shapes:', X_train.shape, X_val.shape, X_test.shape)
```

```python
# ══ 5b. Hyperparameter search — Ridge / Lasso / ElasticNet / SVR ═════════════
#
# Each model is wrapped in a Pipeline(scaler, model) and evaluated with
# RandomizedSearchCV (5-fold CV, neg_MAE scoring).  Best model selected.
#
from sklearn.pipeline import Pipeline

scaler = StandardScaler()

search_configs = [
    {
        'name': 'Ridge',
        'pipeline': Pipeline([('sc', StandardScaler()), ('m', Ridge())]),
        'params':   {'m__alpha': loguniform(1e-2, 1e3)},
        'n_iter':   30
    },
    {
        'name': 'Lasso',
        'pipeline': Pipeline([('sc', StandardScaler()), ('m', Lasso(max_iter=5000))]),
        'params':   {'m__alpha': loguniform(1e-3, 1e2)},
        'n_iter':   30
    },
    {
        'name': 'ElasticNet',
        'pipeline': Pipeline([('sc', StandardScaler()), ('m', ElasticNet(max_iter=5000))]),
        'params':   {'m__alpha': loguniform(1e-3, 1e2), 'm__l1_ratio': uniform(0.05, 0.90)},
        'n_iter':   40
    },
    {
        'name': 'SVR',
        'pipeline': Pipeline([('sc', StandardScaler()), ('m', SVR())]),
        'params':   {'m__C': loguniform(0.1, 200), 'm__epsilon': loguniform(0.01, 20),
                     'm__kernel': ['rbf', 'linear']},
        'n_iter':   25
    },
]

def eval_metrics(name, model, Xv, yv, Xt, yt):
    pv = model.predict(Xv)
    pt = model.predict(Xt)
    rv = dict(MAE=mean_absolute_error(yv,pv), RMSE=mean_squared_error(yv,pv,squared=False), R2=r2_score(yv,pv))
    rt = dict(MAE=mean_absolute_error(yt,pt), RMSE=mean_squared_error(yt,pt,squared=False), R2=r2_score(yt,pt))
    print(f'  {name:12s}  val  MAE={rv["MAE"]:6.2f}  RMSE={rv["RMSE"]:6.2f}  R2={rv["R2"]:+.3f}')
    print(f'  {name:12s}  test MAE={rt["MAE"]:6.2f}  RMSE={rt["RMSE"]:6.2f}  R2={rt["R2"]:+.3f}')
    return rv, rt

results = []
best_val_mae = float('inf')
best_sklearn_model = None
best_sklearn_name  = ''

for cfg in search_configs:
    print(f'\n── {cfg["name"]} HPO ──────────────────────────────')
    rs = RandomizedSearchCV(
        cfg['pipeline'], cfg['params'],
        n_iter=cfg['n_iter'], cv=5,
        scoring='neg_mean_absolute_error',
        random_state=SEED, n_jobs=-1, verbose=0
    )
    rs.fit(X_train, y_train)
    print(f'  Best params: {rs.best_params_}')
    print(f'  CV MAE (5-fold): {-rs.best_score_:.2f}')
    rv, rt = eval_metrics(cfg['name'], rs.best_estimator_, X_val, y_val, X_test, y_test)
    results.append({'model': cfg['name'], 'best_estimator': rs.best_estimator_,
                    'val_MAE': rv['MAE'], 'test_MAE': rt['MAE'], 'val_R2': rv['R2']})
    if rv['MAE'] < best_val_mae:
        best_val_mae          = rv['MAE']
        best_sklearn_model    = rs.best_estimator_
        best_sklearn_name     = cfg['name']

print(f'\n✓  Best sklearn model by val MAE: {best_sklearn_name} (MAE={best_val_mae:.2f})')

# Summary table
pd.DataFrame([{k: v for k,v in r.items() if k != 'best_estimator'} for r in results])
```

```python
# ══ 6. Fine-tune EfficientNet regression head with early stopping ══════════════
#
# Only the newly added classifier head is trained first (3 warm-up epochs),
# then the last two backbone blocks are unfrozen for 7 more epochs.
#
model_ft = models.efficientnet_b0(pretrained=True)

# Replace classifier
try:
    in_feat = model_ft.classifier[1].in_features
except Exception:
    in_feat = [m.in_features for m in model_ft.modules() if isinstance(m, nn.Linear)][-1]

model_ft.classifier = nn.Sequential(
    nn.Dropout(0.3),
    nn.Linear(in_feat, 256),
    nn.SiLU(),
    nn.Dropout(0.2),
    nn.Linear(256, 64),
    nn.SiLU(),
    nn.Linear(64, 1)
)

# Freeze backbone initially
for p in model_ft.features.parameters(): p.requires_grad = False

model_ft = model_ft.to(device)
loss_fn  = nn.HuberLoss(delta=15.0)   # robust to large-depth outliers

def run_val(loader):
    model_ft.eval()
    preds, gts = [], []
    with torch.no_grad():
        for imgs, labels in loader:
            out = model_ft(imgs.to(device)).cpu().numpy().reshape(-1)
            preds.append(out); gts.append(labels.numpy())
    p = np.concatenate(preds); g = np.concatenate(gts)
    return mean_absolute_error(g, p), p, g

history = []

def train_phase(label, epochs, lr, unfreeze_last_n=0):
    if unfreeze_last_n > 0:
        blocks = list(model_ft.features)
        for blk in blocks[-unfreeze_last_n:]:
            for p in blk.parameters(): p.requires_grad = True
    opt = torch.optim.AdamW(
        filter(lambda p: p.requires_grad, model_ft.parameters()),
        lr=lr, weight_decay=1e-4
    )
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    best_mae, patience, patience_limit = float('inf'), 0, 4
    best_state = None
    for ep in range(1, epochs+1):
        model_ft.train()
        ep_loss = []
        for imgs, labels in tqdm(train_loader, desc=f'[{label}] Ep {ep}/{epochs}', leave=False):
            imgs   = imgs.to(device)
            labels = labels.to(device).unsqueeze(1)
            opt.zero_grad()
            loss = loss_fn(model_ft(imgs), labels.float())
            loss.backward(); opt.step()
            ep_loss.append(loss.item())
        sched.step()
        val_mae, _, _ = run_val(val_loader)
        avg_loss = np.mean(ep_loss)
        print(f'[{label}] Ep {ep:02d}  loss={avg_loss:.3f}  val_MAE={val_mae:.2f} cm')
        history.append({'phase': label, 'epoch': ep, 'loss': avg_loss, 'val_mae': val_mae})
        if val_mae < best_mae:
            best_mae   = val_mae
            best_state = {k: v.cpu().clone() for k, v in model_ft.state_dict().items()}
            patience   = 0
        else:
            patience += 1
            if patience >= patience_limit:
                print(f'  → Early stop (patience={patience_limit})')
                break
    if best_state:
        model_ft.load_state_dict(best_state)
    return best_mae

# Phase 1: head-only warm-up
mae1 = train_phase('head-warmup', epochs=5, lr=1e-3)
print(f'Phase-1 best val MAE: {mae1:.2f} cm')

# Phase 2: unfreeze last 2 backbone blocks
mae2 = train_phase('backbone-ft', epochs=10, lr=2e-4, unfreeze_last_n=2)
print(f'Phase-2 best val MAE: {mae2:.2f} cm')
```

```python
# ══ 7. Final evaluation and residual plots ════════════════════════════════════
_, val_preds, val_gts   = run_val(val_loader)
_, test_preds, test_gts = run_val(test_loader)

def report(tag, preds, gts):
    mae  = mean_absolute_error(gts, preds)
    rmse = mean_squared_error(gts, preds, squared=False)
    r2   = r2_score(gts, preds)
    print(f'{tag:8s}  MAE={mae:.2f} cm  RMSE={rmse:.2f} cm  R2={r2:+.3f}')
    return mae, rmse, r2

print('=== EfficientNet fine-tuned ===')
report('val',  val_preds,  val_gts)
report('test', test_preds, test_gts)

print('\n=== Best sklearn baseline ===')
eval_metrics(best_sklearn_name, best_sklearn_model, X_val, y_val, X_test, y_test)

# Training curve
hist_df = pd.DataFrame(history)
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(hist_df['val_mae'].values, marker='o', label='val MAE (cm)')
ax.set_xlabel('Epoch (across phases)'); ax.set_ylabel('MAE (cm)')
ax.set_title('Training curve — validation MAE'); ax.legend()
plt.tight_layout(); plt.show()

# Residual plot
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for ax, preds, gts, tag in [(axes[0], val_preds, val_gts, 'val'),
                              (axes[1], test_preds, test_gts, 'test')]:
    ax.scatter(gts, preds, alpha=0.6, s=20)
    mn, mx = min(gts.min(), preds.min()), max(gts.max(), preds.max())
    ax.plot([mn,mx],[mn,mx],'r--')
    ax.set_xlabel('Actual depth (cm)'); ax.set_ylabel('Predicted depth (cm)')
    ax.set_title(f'{tag} — predicted vs actual')
plt.tight_layout(); plt.show()
```
