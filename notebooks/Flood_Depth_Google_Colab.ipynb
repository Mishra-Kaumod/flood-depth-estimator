# Flood Depth training (Colab)
This notebook prepares the training data, checks label/image alignment and depth distribution, creates stratified 600/50/50 splits, applies simple re-balancing if needed, trains Ridge/Lasso baselines on EfficientNet features, fine-tunes a small EfficientNet regression head, evaluates metrics, and saves a model artifact for download.

```python
import os
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeCV, LassoCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import torch
from torchvision import transforms, models
from PIL import Image
from tqdm import tqdm

# Paths (repo-relative)
DATA_DIR = Path('training_data')
IMAGES_DIR = DATA_DIR / 'images'
LABELS_CSV = DATA_DIR / 'labels.csv'

print('DATA_DIR', DATA_DIR)
print('IMAGES_DIR exists?', IMAGES_DIR.exists())
print('LABELS_CSV exists?', LABELS_CSV.exists())

df = pd.read_csv(LABELS_CSV)
print('Total rows in labels.csv:', len(df))
df.head()
```

```python
# Quick sanity checks: missing files and extra files
missing = []
for fn in df['filename']:
    if not (IMAGES_DIR / fn).exists():
        missing.append(fn)
print('Missing files from training_data/images referenced by labels.csv:', len(missing))
if len(missing):
    print(missing[:20])

extra_files = [p.name for p in IMAGES_DIR.iterdir() if p.is_file() and p.name not in set(df['filename'])] if IMAGES_DIR.exists() else []
print('Extra files in training_data/images not referenced by labels.csv:', len(extra_files))
if len(extra_files):
    print(extra_files[:20])
```

```python
# Create depth bins and visualize distribution
bins = [-1, 5, 20, 50, 80, 1e9]
labels = ['0-5','5-20','20-50','50-80','80+']
df['depth_cm'] = df['depth_cm'].astype(float)
df['bin'] = pd.cut(df['depth_cm'], bins=bins, labels=labels)
counts = df['bin'].value_counts().reindex(labels)
print('Counts per bin (0-5,5-20,20-50,50-80,80+):')
print(counts)
counts.plot(kind='bar', title='Depth distribution (bins)')
plt.show()
```

```python
# Create deterministic stratified splits (target 600/50/50 when dataset has >=700 samples)
seed = 42
n = len(df)
if n >= 700:
    # stratify by depth bin to preserve distribution initially
    train, temp = train_test_split(df, train_size=600, stratify=df['bin'], random_state=seed)
    val, test = train_test_split(temp, train_size=50, test_size=50, stratify=temp['bin'], random_state=seed)
else:
    # fallback proportional split
    train, temp = train_test_split(df, train_size=0.8, stratify=df['bin'], random_state=seed)
    val, test = train_test_split(temp, test_size=0.5, stratify=temp['bin'], random_state=seed)

print('Train/Val/Test sizes:', len(train), len(val), len(test))
print(train['bin'].value_counts())
```

```python
# Rebalance the training set by oversampling each bin to roughly equal counts (if needed)
target_train_size = 600 if len(df) >= 700 else max( int(len(train) * 1.0), len(train))
unique_bins = train['bin'].unique()
nbins = len(labels)
per_bin = max(1, int(target_train_size / nbins))
dfs = []
for lbl in labels:
    grp = train[train['bin'] == lbl]
    if len(grp) == 0:
        continue
    if len(grp) < per_bin:
        extra = grp.sample(per_bin - len(grp), replace=True, random_state=seed)
        grp = pd.concat([grp, extra])
    else:
        grp = grp.sample(per_bin, replace=False, random_state=seed)
    dfs.append(grp)
train_bal = pd.concat(dfs).sample(frac=1, random_state=seed).reset_index(drop=True)
print('Balanced train distribution:')
print(train_bal['bin'].value_counts())
```

```python
# PyTorch dataset + dataloaders (EfficientNet input size 224)
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
])

class FloodDataset(torch.utils.data.Dataset):
    def __init__(self, df, img_dir, transform):
        self.df = df.reset_index(drop=True)
        self.img_dir = Path(img_dir)
        self.transform = transform
    def __len__(self):
        return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        path = self.img_dir / row['filename']
        img = Image.open(path).convert('RGB')
        return self.transform(img), np.float32(row['depth_cm'])

train_ds = FloodDataset(train_bal, IMAGES_DIR, transform)
val_ds = FloodDataset(val, IMAGES_DIR, transform)
test_ds = FloodDataset(test, IMAGES_DIR, transform)

train_loader = torch.utils.data.DataLoader(train_ds, batch_size=32, shuffle=True, num_workers=2)
val_loader = torch.utils.data.DataLoader(val_ds, batch_size=64, shuffle=False, num_workers=2)
test_loader = torch.utils.data.DataLoader(test_ds, batch_size=64, shuffle=False, num_workers=2)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
print('Train/Val/Test lengths:', len(train_ds), len(val_ds), len(test_ds))
```

```python
# Feature extraction using EfficientNet-B0 backbone (pretrained)
backbone = models.efficientnet_b0(pretrained=True)
feature_extractor = torch.nn.Sequential(backbone.features, torch.nn.AdaptiveAvgPool2d(1), torch.nn.Flatten(1)).to(device)
feature_extractor.eval()

def extract_feats(loader):
    Xs = []
    ys = []
    with torch.no_grad():
        for imgs, labels in tqdm(loader):
            imgs = imgs.to(device)
            f = feature_extractor(imgs)
            Xs.append(f.cpu().numpy())
            ys.append(labels.numpy())
    X = np.vstack(Xs)
    y = np.concatenate(ys)
    return X, y

X_train, y_train = extract_feats(train_loader)
X_val, y_val = extract_feats(val_loader)
X_test, y_test = extract_feats(test_loader)
print('Feature shapes:', X_train.shape, X_val.shape, X_test.shape)
```

```python
# Ridge and Lasso baseline on extracted features
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_val_s = scaler.transform(X_val)
X_test_s = scaler.transform(X_test)

alphas = np.logspace(-2, 2, 10)
ridge = RidgeCV(alphas=alphas, cv=5).fit(X_train_s, y_train)
from sklearn.linear_model import Lasso
lasso = Lasso(alpha=0.1, max_iter=5000).fit(X_train_s, y_train)

def report(name, model, Xs, ys):
    pred = model.predict(Xs)
    mae = mean_absolute_error(ys, pred)
    rmse = mean_squared_error(ys, pred, squared=False)
    r2 = r2_score(ys, pred)
    print(f'{name} MAE={mae:.2f} RMSE={rmse:.2f} R2={r2:.3f}')
    return mae, rmse, r2

print('Ridge best alpha:', ridge.alpha_)
report('Ridge val', ridge, X_val_s, y_val)
report('Ridge test', ridge, X_test_s, y_test)
report('Lasso val', lasso, X_val_s, y_val)
report('Lasso test', lasso, X_test_s, y_test)
```

```python
# Fine-tune a small EfficientNet regression head (classifier replaced with regression head)
model_ft = models.efficientnet_b0(pretrained=True)
# classifier structure in torchvision: [Dropout, Linear] or similar - check in runtime
try:
    in_features = model_ft.classifier[1].in_features
except Exception:
    # fallback: inspect last linear layer dynamically
    import inspect
    for n,m in model_ft.named_modules():
        if isinstance(m, torch.nn.Linear):
            in_features = m.in_features
            break

model_ft.classifier = torch.nn.Sequential(
    torch.nn.Dropout(0.2),
    torch.nn.Linear(in_features, 256),
    torch.nn.ReLU(),
    torch.nn.Dropout(0.1),
    torch.nn.Linear(256, 1)
)
model_ft = model_ft.to(device)
optimizer = torch.optim.Adam(model_ft.classifier.parameters(), lr=1e-3)
loss_fn = torch.nn.SmoothL1Loss()

epochs = 5
for epoch in range(epochs):
    model_ft.train()
    loop = tqdm(train_loader, desc=f'Epoch {epoch+1}/{epochs}')
    for imgs, labels in loop:
        imgs = imgs.to(device)
        labels = labels.to(device).unsqueeze(1).float()
        preds = model_ft(imgs)
        loss = loss_fn(preds, labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    # validation MAE after epoch
    model_ft.eval()
    preds = []
    gts = []
    with torch.no_grad():
        for imgs, labels in val_loader:
            imgs = imgs.to(device)
            out = model_ft(imgs).cpu().numpy().reshape(-1)
            preds.append(out)
            gts.append(labels.numpy())
    preds = np.concatenate(preds)
    gts = np.concatenate(gts)
    print(f'After epoch {epoch+1} val MAE:', mean_absolute_error(gts, preds))

# final test eval
model_ft.eval()
preds = []
gts = []
with torch.no_grad():
    for imgs, labels in test_loader:
        imgs = imgs.to(device)
        out = model_ft(imgs).cpu().numpy().reshape(-1)
        preds.append(out)
        gts.append(labels.numpy())
preds = np.concatenate(preds)
gts = np.concatenate(gts)
print('Final test MAE:', mean_absolute_error(gts, preds), 'RMSE:', mean_squared_error(gts, preds, squared=False), 'R2:', r2_score(gts, preds))
```

```python
# Save trained PyTorch model and baseline artifacts for download
OUT_DIR = Path('models')
OUT_DIR.mkdir(exist_ok=True)
OUT_PATH = OUT_DIR / 'colab_trained_model.pth'
torch.save({
    'model_state_dict': model_ft.state_dict(),
    'scaler': scaler,
    'ridge': ridge
}, str(OUT_PATH))
print('Saved:', OUT_PATH)
# In Colab you can download with files.download; try here and fallback if not available
try:
    from google.colab import files
    files.download(str(OUT_PATH))
except Exception as e:
    print('files.download not available in this environment; file saved to', OUT_PATH)
```
